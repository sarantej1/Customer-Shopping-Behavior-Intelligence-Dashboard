```python
sales['Review Rating'] = sales.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
sales.columns = sales.columns.str.lower()
sales.columns = sales.columns.str.replace(' ', '_')
sales = sales.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
labels = ['Young Adult', 'Adult', 'Middle Aged', 'Senior']
sales['age_group'] = pd.qcut(sales['age'], q = 4, labels = labels)
```

```python
sales[['age','age_group']]
```

```python
sales['frequency_of_purchases']
```

```python
sales['frequency_of_purchases'].value_counts()
```

```python
maps = {  'Every 3 Months' : 90,
'Annually' : 365,
'Quarterly' : 90,
'Monthly'	: 30,
'Bi-Weekly'	: 14,
'Fortnightly'	: 14,
'Weekly':7       }

sales['Purchase_frequency_days']= sales['frequency_of_purchases'].map(maps)
```

```python
sales[['frequency_of_purchases', 'Purchase_frequency_days']]
```

```python
sales.head()
```

```python
(sales['discount_applied']==sales['promo_code_used']).all()
```

```python
sales = sales.drop('promo_code_used', axis = 1)
```

```python
sales.head()
```

```python
revenue_by_gender = sales.groupby('gender')['purchase_amount'].sum().reset_index()
display(revenue_by_gender)

sales = sales.merge(revenue_by_gender.rename(columns={'purchase_amount': 'total_revenue_by_gender'}), on='gender', how='left')
display(sales[['gender', 'purchase_amount', 'total_revenue_by_gender']].head())
```

```python
avg_purchase = sales['purchase_amount'].mean()
high_spenders_with_discount = sales[(sales['discount_applied'] == 'Yes') & (sales['purchase_amount'] > avg_purchase)]

print(f"Average Purchase Amount: {avg_purchase:.2f}")
high_spenders_with_discount[['customer_id', 'purchase_amount']]
```

```python
top_rated_products = sales.groupby('item_purchased')['review_rating'].mean().sort_values(ascending=False).head(5).reset_index()
display(top_rated_products)
```

```python
sales.head()
```

```python
avg_standard = sales[sales['shipping_type'] == 'Standard']['purchase_amount'].mean()
avg_express = sales[sales['shipping_type'] == 'Express']['purchase_amount'].mean()

shipping_comparison = pd.DataFrame({
    'Shipping Type': ['Standard', 'Express'],
    'Average Purchase Amount': [avg_standard, avg_express]
})

shipping_comparison
```

```python
sales['subscription_status'].value_counts()
```

```python
sub_comparison = sales.groupby('subscription_status')['purchase_amount'].agg(['mean', 'sum']).reset_index()
sub_comparison.columns = ['Subscription Status', 'Average Spend', 'Total Revenue']
display(sub_comparison)
```

```python
discount_counts = sales[sales['discount_applied'] == 'Yes'].groupby('item_purchased').size()
total_counts = sales.groupby('item_purchased').size()
discount_percentage = (discount_counts / total_counts * 100).sort_values(ascending=False).head(5).reset_index()
discount_percentage.columns = ['Item Purchased', 'Discount Applied %']
display(discount_percentage)
```

```python
sales.head()
```

```python
def segment_customer(purchases):
    if purchases ==1:
        return 'New'
    elif 2 <= purchases <= 10 :
        return 'Returning'
    else:
        return 'Loyal'

sales['customer_segment'] = sales['previous_purchases'].apply(segment_customer)
segment_counts = sales['customer_segment'].value_counts().reset_index()
segment_counts.columns = ['Customer Segment', 'Count']
display(segment_counts)
```

```python
items = sales.groupby(['category', 'item_purchased']).size().reset_index(name='total_count')
items = items.sort_values(by='total_count', ascending=False).head(3).copy()
items['item_rank'] = range(1, 4)
display(items[['item_rank', 'item_purchased', 'category', 'total_count']])
```

```python
def purchases(row):
  if row['subscription_status'] == 'Yes' and row['previous_purchases'] > 5:
    return 'Yes'
  else:
    return 'No'

subs = sales.apply(purchases, axis=1)
subs.value_counts()
```

```python
agegroup = sales.groupby('age_group', observed=False)['purchase_amount'].sum().sort_values(ascending=False).reset_index()
agegroup.columns = ['Age Group', 'Total Revenue']
display(agegroup)
```
